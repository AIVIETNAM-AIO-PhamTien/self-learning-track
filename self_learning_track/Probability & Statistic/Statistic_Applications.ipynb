```python
### Question 4

def compute_correlation_cofficient(X, Y):
  N = len(X)
  # Your code here #
  # Tính tử số

  numerator = N * sum(X[i]*Y[i] for i in range(N)) - sum(X)*sum(Y)

  # Tính mẫu
  # Tổng X bình phương, Y bình phương
  sum_square_X = sum(x**2 for x in X)
  sum_square_Y = sum(y**2 for y in Y)

  # Bình phương của tổng X, Y
  square_sum_X = sum(X)**2
  square_sum_Y = sum(Y)**2

  denominator = np.sqrt(N * sum_square_X - square_sum_X) * np.sqrt(N * sum_square_Y - square_sum_Y)

  return np.round(numerator / denominator,2)


X = np.asarray([-2, -5, -11, 6, 4, 15, 9])
Y = np.asarray([4, 25, 121, 36, 16, 225, 81])
print("Correlation: ", compute_correlation_cofficient(X,Y))

# Câu D
```

```python
def correlation(x, y):
  x_mean = np.mean(x)
  y_mean = np.mean(y)

  N = len(x)

  # Your code here #
  numerator = sum((x[i] - x_mean) * (y[i] - y_mean) for i in range(N))

  denominator = (
      np.sqrt(sum((xi - x_mean) ** 2 for xi in x)) *
      np.sqrt(sum((yi - y_mean)** 2 for yi in y))
      )


  return np.round(numerator / denominator,2)





```

## Plagiarism Checker

```python
docs = [
  "Học máy là một nhánh của trí tuệ nhân tạo",
  "Trí tuệ nhân tạo bao gồm học máy và mạng nơ ron",
  "Mạng nơ ron là một mô hình quan trọng trong học sâu",
  "Học sâu là một lĩnh vực của trí tuệ nhân tạo và học máy"
]
```

```python
# 13. Tạo danh sách từ vựng (vocabulary) bằng cách hoàn thiện hàm build_vocabulary

def build_vocabulary(docs):
  vocab_set = set()
  for doc in docs:
    words = doc.lower().split()
    vocab_set.update(words)
  vocab = sorted(list(vocab_set)) #sort trước khi trả theo yêu cầu
  return vocab
# Câu B
```

```python
# Câu 14: Tính TF (Term Frequency)

def compute_tf(doc, vocab):
  words = doc.lower().split()
  tf = np.zeros(len(vocab))
  for i, term in enumerate(vocab):
    tf[i] = words.count(term) / len(words)   # tf = số lần từ xuất hiện / tổng số từ
  return tf

# Câu A
```

```python
# Câu 15 Tính IDF (Inverse Document Frequency)

def compute_idf(docs, vocab):
  N = len(docs)
  idf = np.zeros(len(vocab))
  for i, term in enumerate(vocab):
     df = sum([1 for doc in docs if term in doc.lower().split()])
     idf[i] = np.log((N + 1) / (df + 1)) + 1   # Tổng số tài liệu / Tổng số tài liệu xuất hiện từ, có smooth
  return idf

# Câu C
```

```python
# Câu 16 :  Tính TF-IDF vector

def compute_tfidf(tf, idf):
  tf_idf = tf * idf    # Bằng tích tf và idf
  return tf_idf

# Câu C
```

```python
# câu 17: Tính độ tương đồng Cosine Similarity

def cosine_similarity(vec1, vec2):
  dot_product = np.dot(vec1, vec2)
  norm1 = np.linalg.norm(vec1)
  norm2 = np.linalg.norm(vec2)
  return dot_product / (norm1 * norm2)
# Câu A
```

```python
# câu 18 Hoàn thiện dòng vẽ heatmap

docs = [
  "Học máy là một nhánh của trí tuệ nhân tạo",
  "Trí tuệ nhân tạo bao gồm học máy và mạng nơ ron",
  "Mạng nơ ron là một mô hình quan trọng trong học sâu",
  "Học sâu là một lĩnh vực của trí tuệ nhân tạo và học máy"
]

def build_vocabulary(docs):
  vocab_set = set()
  for doc in docs:
    vocab_set.update(doc)
  vocab = sorted(list(vocab_set)) #sort trước khi trả theo yêu cầu
  return vocab

def preprocess(text):
    text = text.lower()  # Chuyển về chữ thường
    words = text.split()  # Tách từ
    return words

def create_corpus(docs):
    """Tạo corpus từ danh sách văn bản"""
    processed_docs = [preprocess(doc) for doc in docs]
    vocabulary = build_vocabulary(processed_docs)
    return processed_docs, vocabulary

def compute_tf(word_list, vocabulary):
    """Tính TF (Term Frequency)"""
    word_count = {}
    for word in word_list:
        word_count[word] = word_count.get(word, 0) + 1
    tf_vector = np.array([word_count.get(word, 0) / len(word_list) for word in vocabulary]) # Hàm get lấy giá trị của wordm k có trả 0
    return tf_vector

def compute_idf(docs, vocabulary):
    """Tính IDF (Inverse Document Frequency)"""
    num_docs = len(docs)
    idf_vector = np.zeros(len(vocabulary))
    for i, term in enumerate(vocabulary):
      df = sum(1 for doc in docs if term in doc)
      idf_vector[i] = np.log((num_docs + 1) / (df + 1)) + 1   # Tổng số tài liệu / Tổng số tài liệu xuất hiện từ, có smooth
    return idf_vector

def compute_tf_idf(docs):
    """Tính ma trận TF-IDF"""
    # Tạo corpus
    processed_docs, vocabulary = create_corpus(docs)

    # Tính TF và IDF
    tf_matrix = np.array([compute_tf(doc, vocabulary) for doc in processed_docs])
    idf_vector = compute_idf(processed_docs, vocabulary)

    # Tính TF-IDF
    tf_idf_matrix = tf_matrix * idf_vector
    return tf_idf_matrix

def cosine_similarity(vec1, vec2):
    """Tính độ tương đồng cosine giữa hai vector"""
    dot_product = np.dot(vec1, vec2)
    norm1 = np.linalg.norm(vec1)
    norm2 = np.linalg.norm(vec2)
    return dot_product / (norm1 * norm2)

def check_plagiarism_all(docs):
    """Tính độ tương đồng giữa tất cả các cặp văn bản"""
    tf_idf_matrix = compute_tf_idf(docs)
    num_docs = len(docs)

    similarity_matrix = np.zeros((num_docs, num_docs))

    for i in range(num_docs):
        for j in range(num_docs):
            similarity_matrix[i, j] = cosine_similarity(tf_idf_matrix[i], tf_idf_matrix[j])

    return similarity_matrix


similarity_matrix = check_plagiarism_all(docs)

plt.figure(figsize=(8, 6))
sns. heatmap (similarity_matrix , annot =True , fmt='.3f', cmap ='YlOrRd')
plt.title('Ma trận độ tương đồng Cosine')
plt.show()
```
